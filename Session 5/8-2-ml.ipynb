```python
from tensorflow.keras.datasets import mnist
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train_flat = x_train.reshape(60000, 28 * 28)
x_test_flat = x_test.reshape(10000, 28 * 28)
x_train_flat.shape
```

```python

# 3) تطبيع القيم
# بدل أن تكون القيم من 0 إلى 255 تصبح من 0 إلى 1
x_train_flat = x_train_flat / 255.0
x_test_flat = x_test_flat / 255.0

# 4) إنشاء النموذج
model = LogisticRegression(max_iter=1000)

# 5) تدريب النموذج
model.fit(x_train_flat, y_train)

# 6) تقييم النموذج
y_pred = model.predict(x_test_flat)
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python

# 7) تجربة صورة جديدة
index = 1

new_image = x_test[index]
new_image_flat = new_image.reshape(1, 28 * 28) / 255.0

prediction = model.predict(new_image_flat)

print("Predicted digit:", prediction[0])
print("Real digit:", y_test[index])
```

```python

# 8) عرض الصورة
plt.imshow(new_image, cmap="gray")
plt.title(f"Predicted: {prediction[0]} | Real: {y_test[index]}")
plt.axis("off")
plt.show()
```
