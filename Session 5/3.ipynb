```python
import cv2
import numpy as np

img = cv2.imread('image.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# تطبيق Harris
corners = cv2.cornerHarris(gray, 2, 3, 0.04)

# تحديد الزوايا
img[corners > 0.01 * corners.max()] = [0, 0, 255]

cv2.imshow('Corners', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
