```python
from skimage.feature import local_binary_pattern
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('image.jpg', 0)

# تطبيق LBP
lbp = local_binary_pattern(img, 8, 1, method='uniform')

plt.imshow(lbp, cmap='gray')
plt.title("LBP")
plt.axis('off')
plt.show()
```
