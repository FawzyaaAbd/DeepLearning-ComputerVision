```python
import cv2

img = cv2.imread('person.jpg', 0)

orb = cv2.ORB_create()

# استخراج النقاط
keypoints, descriptors = orb.detectAndCompute(img, None)

# رسم النقاط
img2 = cv2.drawKeypoints(img, keypoints, None)

cv2.imshow('ORB Features', img2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
