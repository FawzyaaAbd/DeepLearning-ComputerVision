```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Dense, Lambda
from tensorflow.keras.utils import to_categorical
import tensorflow.keras.backend as K

# -----------------------------------
# 1) قاموس كلمات بسيط
# -----------------------------------
word_to_index = {
    "i": 0,
    "love": 1,
    "deep": 2,
    "learning": 3
}

index_to_word = {v: k for k, v in word_to_index.items()}
vocab_size = len(word_to_index)
```

```python
vocab_size
```

```python

# -----------------------------------
# 2) بيانات CBOW جاهزة (بدون Tokenizer)
# -----------------------------------
# Context → Target
X = np.array([
    [word_to_index["i"], word_to_index["deep"]],        # i, deep → love
    [word_to_index["love"], word_to_index["learning"]]  # love, learning → deep
])
X
```

```python
word_to_index["deep"]
```

```python

y = np.array([
    word_to_index["love"],
    word_to_index["deep"]
])

# تحويل target إلى one-hot
y = to_categorical(y, num_classes=vocab_size)
y
```

```python

# -----------------------------------
# 3) النموذج
# -----------------------------------
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=5, input_length=2),
    
    # نأخذ متوسط الكلمتين
    Lambda(lambda x: K.mean(x, axis=1)),
    
    Dense(vocab_size, activation='softmax')
])

model.compile(loss='categorical_crossentropy', optimizer='adam')

# -----------------------------------
# 4) التدريب
# -----------------------------------
model.fit(X, y, epochs=500, verbose=0)
```

```python

# -----------------------------------
# 5) اختبار
# -----------------------------------
def predict(w1, w2):
    x = np.array([[word_to_index[w1], word_to_index[w2]]])
    pred = model.predict(x, verbose=0)
    return index_to_word[np.argmax(pred)]

print(predict("i", "deep"))       # متوقع: love
print(predict("love", "learning")) # متوقع: deep
```
