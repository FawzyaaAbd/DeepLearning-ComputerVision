```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding

# -----------------------------------
# 1) تعريف الكلمات
# -----------------------------------
words = ["king", "queen", "man", "woman"]

word_to_index = {word: i for i, word in enumerate(words)}
index_to_word = {i: word for word, i in word_to_index.items()}

vocab_size = len(words)

# -----------------------------------
# 2) إنشاء نموذج Embedding
# -----------------------------------
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=6)
])

# -----------------------------------
# 3) بناء النموذج (حل المشكلة 🔥)
# -----------------------------------
model.build(input_shape=(None, 1))

# -----------------------------------
# 4) استخراج الـ embeddings
# -----------------------------------
embeddings = model.get_weights()[0]

def get_vector(word):
    return embeddings[word_to_index[word]]
```

```python
get_vector("man")
```

```python

# -----------------------------------
# 5) العملية الرياضية
# king - man + woman
# -----------------------------------
result_vector = get_vector("king") - get_vector("man") + get_vector("woman")

# -----------------------------------
# 6) إيجاد أقرب كلمة
# -----------------------------------
def closest_word(vector):
    max_sim = -1
    best_word = None
    
    for word in words:
        vec = get_vector(word)
        
        # حساب cosine similarity
        sim = np.dot(vector, vec) / (np.linalg.norm(vector) * np.linalg.norm(vec))
        
        if sim > max_sim:
            max_sim = sim
            best_word = word
            
    return best_word

# -----------------------------------
# 7) النتيجة
# -----------------------------------
print("Result:", closest_word(result_vector))
```
