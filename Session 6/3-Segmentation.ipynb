<h2 style="color:#006064;">التجزئة باستخدام SLIC — البكسلات الفائقة (Superpixels)</h2>

<p><strong>SLIC (Simple Linear Iterative Clustering)</strong> هو خوارزمية تجزئة تُقسّم الصورة إلى مناطق صغيرة ومتماسكة تُسمى <strong>Superpixels</strong>، وهي أكبر من البكسل الواحد لكن أصغر من الكائنات في الصورة.</p>

<p>تعتمد SLIC على نسخة معدَّلة من K-Means تعمل في فضاء <strong>CIELAB للألوان</strong> مضاف إليه <strong>الموضع المكاني</strong>، مما يجعل كل Superpixel متماسكاً لونياً وشكلياً.</p>

<table style="width:100%; border-collapse:collapse; text-align:right; direction:rtl; margin-top:10px;">
  <tr style="background:#b2ebf2;">
    <th style="padding:8px; border:1px solid #4dd0e1;">المعامل</th>
    <th style="padding:8px; border:1px solid #4dd0e1;">الوصف</th>
  </tr>
  <tr><td style="padding:8px; border:1px solid #4dd0e1;"><code>n_segments</code></td><td style="padding:8px; border:1px solid #4dd0e1;">العدد التقريبي للـ Superpixels</td></tr>
  <tr><td style="padding:8px; border:1px solid #4dd0e1;"><code>compactness</code></td><td style="padding:8px; border:1px solid #4dd0e1;">الموازنة بين اللون والموضع — قيمة أعلى = شكل أكثر انتظاماً</td></tr>
  <tr><td style="padding:8px; border:1px solid #4dd0e1;"><code>sigma</code></td><td style="padding:8px; border:1px solid #4dd0e1;">مقدار التمويه الأولي لتقليل الضجيج</td></tr>

<p style="margin-top:12px;"><strong>الفائدة:</strong> تُستخدم Superpixels كمرحلة مسبقة (Pre-processing) قبل خوارزميات التجزئة الأكثر تعقيداً لتقليل عدد الوحدات المعالجة وتسريع التنفيذ.</p>

</div>

```python
from skimage.segmentation import slic, mark_boundaries
from skimage.io import imread
import matplotlib.pyplot as plt

image = imread("image.png")

segments = slic(
    image,
    n_segments=50,
    compactness=10,
    sigma=1,
    start_label=0
)

plt.figure(figsize=(10, 6))
plt.imshow(mark_boundaries(image, segments))
plt.axis("off")
plt.title("SLIC Superpixels")
plt.show()
```
