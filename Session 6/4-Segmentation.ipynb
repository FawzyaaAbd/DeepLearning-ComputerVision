<h2 style="color:#bf360c;">التجزئة باستخدام Felzenszwalb — تجزئة قائمة على الرسم البياني</h2>

<p><strong>خوارزمية Felzenszwalb</strong> هي خوارزمية تجزئة سريعة تعتمد على نظرية الرسوم البيانية (Graph Theory)، حيث تُعامَل الصورة كرسم بياني:</p>
<ul>
  <li>كل <strong>بكسل</strong> هو عقدة (Node) في الرسم البياني</li>
  <li>كل <strong>حافة</strong> (Edge) بين بكسلين متجاورين تحمل وزناً يعتمد على الفرق في شدة اللون</li>
  <li>تُدمج المناطق ذات التغيرات اللونية المتشابهة في مجموعة واحدة</li>
</ul>

<table style="width:100%; border-collapse:collapse; text-align:right; direction:rtl; margin-top:10px;">
  <tr style="background:#ffe0b2;">
    <th style="padding:8px; border:1px solid #ffb74d;">المعامل</th>
    <th style="padding:8px; border:1px solid #ffb74d;">التأثير</th>
  </tr>
  <tr><td style="padding:8px; border:1px solid #ffb74d;"><code>scale</code></td><td style="padding:8px; border:1px solid #ffb74d;">قيمة أعلى = مناطق أكبر وأقل في العدد</td></tr>
  <tr><td style="padding:8px; border:1px solid #ffb74d;"><code>sigma</code></td><td style="padding:8px; border:1px solid #ffb74d;">مقدار التمويه الأولي لتقليل تأثير الضجيج</td></tr>
  <tr><td style="padding:8px; border:1px solid #ffb74d;"><code>min_size</code></td><td style="padding:8px; border:1px solid #ffb74d;">الحجم الأدنى للمنطقة — يمنع ظهور مناطق صغيرة جداً</td></tr>
</table>

<p style="margin-top:12px;"><strong>ميزة الخوارزمية:</strong> لا تحتاج إلى تحديد عدد المجموعات مسبقاً، بل تحدده تلقائياً بناءً على بنية الصورة.</p>

</div>

```python
from skimage.segmentation import felzenszwalb, mark_boundaries
from skimage.io import imread
import matplotlib.pyplot as plt

image = imread("image.png")

segments = felzenszwalb(
    image,
    scale=300,
    sigma=0.8,
    min_size=20
)

plt.figure(figsize=(10, 6))
plt.imshow(mark_boundaries(image, segments))
plt.axis("off")
plt.title("Felzenszwalb Segmentation")
plt.show()
```
