<tr style="background:#fff3e0;"><td style="padding:8px;border:1px solid #ddd;">النسيان</td><td style="padding:8px;border:1px solid #ddd;">ماذا ننسى؟</td><td style="padding:8px;border:1px solid #ddd;">f = sigmoid(Wf[h,x])</td></tr>

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

# -----------------------------
# 1) البيانات
# -----------------------------
X = np.array([
    [1, 2, 3],
    [2, 3, 4],
    [3, 4, 5],
    [4, 5, 6]
], dtype=float)

y = np.array([4, 5, 6, 7], dtype=float)

# تحويل الشكل إلى 3D
X = X.reshape((X.shape[0], X.shape[1], 1))

# -----------------------------
# 2) بناء النموذج
# -----------------------------
model = Sequential([
    LSTM(20, activation='tanh', input_shape=(3, 1)),
    Dense(1)
])

# -----------------------------
# 3) التجهيز
# -----------------------------
model.compile(optimizer='adam', loss='mse')

# -----------------------------
# 4) التدريب
# -----------------------------
model.fit(X, y, epochs=200, verbose=0)

# -----------------------------
# 5) التنبؤ
# -----------------------------
test = np.array([[5, 6, 7]], dtype=float)
test = test.reshape((1, 3, 1))

prediction = model.predict(test, verbose=0)

print("Prediction:", prediction[0][0])
```
