<h1>الشبكات العصبية المتكررة — RNN</h1>
<div>RNN تعالج التسلسلات خطوة بخطوة مع الاحتفاظ بذاكرة داخلية.</div>
<p>المعادلة الأساسية: <code>h_t = tanh(W_x·x_t + W_h·h_{t-1} + b)</code></p>

<h2>مشكلة التدرج المتلاشي — Vanishing Gradient</h2>
<div>عند تدريب RNN على تسلسلات طويلة، تضعف إشارة التحديث كلما رجعنا في الزمن، مما يجعل الشبكة تنسى المعلومات البعيدة ولا تستطيع تعلم التبعيات الطويلة المدى.</div>
</div>

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# -----------------------------
# 1) تجهيز البيانات
# -----------------------------
X = np.array([
    [1, 2, 3],
    [2, 3, 4],
    [3, 4, 5],
    [4, 5, 6]
], dtype=float)

y = np.array([4, 5, 6, 7], dtype=float)


#1,ذهب
#2,الولد
#3,إلى
#4,المدرسة
#5,الحديقة
#6,جميلة

#1,2,3,4,0
#1,2,3,5,0
#4,6,0,0,0
#5,6,0,0,0

# RNN تحتاج الشكل:
# (عدد العينات, عدد الخطوات الزمنية, عدد الخصائص)
X = X.reshape((X.shape[0], X.shape[1], 1))
#X = X.reshape((4, 3, 1))

X

```

```python
X.shape[1]
```

```python
# -----------------------------
# 2) بناء النموذج
# -----------------------------
model = Sequential([
    SimpleRNN(10, activation='tanh', input_shape=(3, 1)),
    Dense(1)
])

# -----------------------------
# 3) تجهيز التدريب
# -----------------------------
model.compile(optimizer='adam', loss='mse')

# -----------------------------
# 4) تدريب النموذج
# -----------------------------
model.fit(X, y, epochs=200, verbose=0)

# -----------------------------
# 5) تجربة التنبؤ
# -----------------------------
test = np.array([[5, 6, 7]], dtype=float)
test = test.reshape((1, 3, 1))
test
```

<h2>ملخص</h2>
<p>RNN جيد للتسلسلات البسيطة — لكنه يعاني مع التبعيات الطويلة بسبب مشكلة التدرج المتلاشي.</p>
<p>الحل: <strong>LSTM</strong> (Long Short-Term Memory) التي تستخدم بوابات ذكية لتذكر المعلومات المهمة على المدى البعيد.</p>
</div>

```python
prediction = model.predict(test, verbose=0)

print("Prediction:", prediction[0][0])
```
