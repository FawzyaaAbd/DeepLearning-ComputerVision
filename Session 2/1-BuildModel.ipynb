```python
import re
import numpy as np
import pandas as pd
from gensim.models import Word2Vec
# 1) تحميل البيانات
df = pd.read_excel("political_sports_articles_arabic_dataset.xlsx")
df.head(3)
```

```python
# 2) دالة تنظيف بسيطة
def preprocess_text(text):
    text = str(text).lower()
    text = re.sub(r'[^\u0600-\u06FF\s]', ' ', text)   # إبقاء العربية والمسافات فقط
    text = re.sub(r'\s+', ' ', text).strip()
    return text.split()
# 3) تجهيز النصوص
sentences = df["text"].apply(preprocess_text).tolist()
```

```python
len(sentences)
len(sentences[1])
```

```python

# 4) تدريب Word2Vec
model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=5,
    min_count=1,
    workers=4,
    sg=1   # 1 = Skip-gram
)

# 5) حفظ المودل
model.save("arabic_word2vec.model")
print("تم تدريب المودل وحفظه بنجاح")
```
