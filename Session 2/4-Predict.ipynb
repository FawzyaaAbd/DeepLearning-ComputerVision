```python
import re
import joblib
import numpy as np
from gensim.models import Word2Vec

# 1) تحميل المودلات
w2v_model = Word2Vec.load("arabic_word2vec.model")
clf_model = joblib.load("logistic_regression_model.pkl")
label_encoder = joblib.load("label_encoder.pkl")
```

```python

# 2) تنظيف النص
def preprocess_text(text):
    text = str(text).lower()
    text = re.sub(r'[^\u0600-\u06FF\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text.split()

# 3) تحويل النص إلى شعاع
def text_to_vector(text, model):
    words = preprocess_text(text)
    word_vectors = [model.wv[word] for word in words if word in model.wv]

    if len(word_vectors) == 0:
        return np.zeros(model.vector_size)

    return np.mean(word_vectors, axis=0)
```

```python

# 4) النص الجديد
new_text = "السياسية"

# 5) تحويل النص إلى شعاع
text_vector = text_to_vector(new_text, w2v_model).reshape(1, -1)
text_vector
```

```python
# 6) التنبؤ
predicted_label_num = clf_model.predict(text_vector)[0]
predicted_label_num
```

```python
predicted_label = label_encoder.inverse_transform([predicted_label_num])[0]

print("الصنف المتوقع:", predicted_label)
```
