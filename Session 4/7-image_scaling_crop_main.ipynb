<div dir="rtl" align="right">
<b>استيراد المكتبات</b><br>
نستورد مكتبة <code>cv2</code> لمعالجة الصور، و<code>matplotlib</code> لعرضها، و<code>numpy</code> للعمليات على المصفوفات.
</div>

```python
import cv2
from matplotlib import pyplot as plt
import numpy as np
```

<div dir="rtl" align="right">
<b>قراءة الصورة الأصلية</b><br>
نحمّل صورة الأزهار من القرص وتُخزَّن في المتغير <code>image</code> بنظام BGR بأبعادها الأصلية 980×980 بكسل.
</div>

```python
image = cv2.imread('flowers.jpg')
```

<div dir="rtl" align="right">
<b>عرض الصورة الأصلية</b><br>
نعرض الصورة الأصلية قبل إجراء أي تغيير على أبعادها كمرجع للمقارنة.
</div>

```python
plt.imshow(cv2.cvtColor(image,cv2.COLOR_BGR2RGB))
plt.show()
```

<div dir="rtl" align="right">
<b>تصغير الصورة (Downscaling)</b><br>
نحدد الأبعاد الجديدة الأصغر (300×200 بكسل) ثم نستخدم <code>cv2.resize()</code> لتقليص الصورة.<br>
الخوارزمية <code>INTER_LINEAR</code> (الاستيفاء الخطي الثنائي) هي الأنسب للتصغير، تحسب قيمة كل بكسل من متوسط البكسلات المجاورة للحصول على نتيجة ناعمة.
</div>

```python
down_width = 300
down_hight = 200
down_points = (down_width,down_hight)
resized_down = cv2.resize(image,down_points,cv2.INTER_LINEAR)
```

<div dir="rtl" align="right">
<b>عرض الصورة المصغّرة</b><br>
نعرض الصورة بعد تصغيرها إلى 300×200 بكسل لملاحظة الفرق في الحجم مقارنةً بالأصل.
</div>

```python
plt.imshow(cv2.cvtColor(resized_down,cv2.COLOR_BGR2RGB))
plt.show()
```

<div dir="rtl" align="right">
<b>تكبير الصورة (Upscaling)</b><br>
نحدد الأبعاد الجديدة الأكبر (1200×400 بكسل) ونستخدم <code>cv2.resize()</code> لتكبير الصورة.<br>
تُستخدم <code>INTER_LINEAR</code> أيضاً هنا لتقدير قيم البكسلات الجديدة بالاستيفاء بين القيم الأصلية.
</div>

```python
up_width = 1200
up_hight = 400
up_points = (up_width,up_hight)
resized_up = cv2.resize(image,up_points,cv2.INTER_LINEAR)
```

<div dir="rtl" align="right">
<b>عرض الصورة المكبّرة</b><br>
نعرض الصورة بعد تكبيرها إلى 1200×400 بكسل. لاحظ أن التكبير الزائد قد يُظهر بعض التشويش لأن المعلومات الجديدة تُحسب بالتقدير لا من بيانات حقيقية.
</div>

```python
plt.imshow(cv2.cvtColor(resized_up,cv2.COLOR_BGR2RGB))
plt.show()
```

<div dir="rtl" align="right">
<b>اقتصاص جزء من الصورة (Cropping)</b><br>
نقتصّ منطقة محددة من الصورة المكبّرة باستخدام فهرسة المصفوفات:<br>
<code>resized_up[150:350, 0:600]</code><br>
الصيغة هي <code>[الصفوف من:إلى، الأعمدة من:إلى]</code> أي من الصف 150 إلى 350 (الارتفاع) ومن العمود 0 إلى 600 (العرض).
</div>

```python
cropped_image = image[80:210, 715:850]
```

<div dir="rtl" align="right">
<b>عرض الصورة المقتصّة</b><br>
نعرض الجزء المقتصّ من الصورة. الاقتصاص لا يعدّل الصورة الأصلية بل ينشئ مصفوفة جديدة تمثل المنطقة المحددة فقط.
</div>

```python
plt.imshow(cv2.cvtColor(cropped_image,cv2.COLOR_BGR2RGB))
plt.show()
```
