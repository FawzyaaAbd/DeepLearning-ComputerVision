<b>استيراد المكتبات</b><br>
نستورد مكتبة <code>cv2</code> لمعالجة الصور، و<code>matplotlib</code> لعرضها، و<code>numpy</code> للعمليات الرياضية على المصفوفات.
</div>

```python
import cv2
from matplotlib import pyplot as plt
import numpy as np
```

<div dir="rtl" align="right">
<b>قراءة الصورة وتحويلها إلى رمادي</b><br>
نقرأ الصورة من الملف ثم نحولها فوراً إلى تدرج رمادي (Grayscale). خوارزمية Canny تعمل على الصور الرمادية لأنها تحتاج إلى حساب التدرجات (Gradients) في شدة الإضاءة فقط.
</div>

```python
image = cv2.imread('flowers.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

<div dir="rtl" align="right">
<b>اكتشاف الحواف باستخدام خوارزمية Canny</b><br>
نستخدم <code>cv2.Canny()</code> لاكتشاف الحواف في الصورة. تعمل الخوارزمية بأربع خطوات: تنعيم الضوضاء، حساب التدرجات، قمع غير القمم (Non-Maximum Suppression)، وأخيراً العتبة المزدوجة (Double Thresholding).<br>
<code>100</code>: الحد الأدنى للعتبة — الحواف تحته تُحذف.<br>
<code>200</code>: الحد الأعلى للعتبة — الحواف فوقه تُقبل بشكل مؤكد.
</div>

```python
edges = cv2.Canny(image, 100,200)
```

<div dir="rtl" align="right">
<b>عرض الصورة الرمادية الأصلية</b><br>
نعرض الصورة بعد تحويلها إلى رمادي كمرجع للمقارنة مع نتيجة اكتشاف الحواف.
</div>

```python
plt.imshow(image,cmap='gray', vmin=0, vmax=255)
plt.show()
```

<div dir="rtl" align="right">
<b>عرض نتيجة اكتشاف الحواف</b><br>
نعرض الصورة الناتجة من خوارزمية Canny. تظهر الحواف باللون الأبيض على خلفية سوداء، حيث يمثل كل بكسل أبيض انتقالاً حاداً في شدة الإضاءة يدل على حافة في الصورة الأصلية.
</div>

```python
plt.imshow(edges,cmap='gray', vmin=0, vmax=255)
plt.show()
```
