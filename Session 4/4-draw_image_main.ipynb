<div dir="rtl" align="right">
<b>استيراد المكتبات</b><br>
نستورد مكتبة <code>numpy</code> لإنشاء المصفوفات، و<code>matplotlib</code> لعرض الصور، و<code>cv2</code> لرسم الأشكال عليها.
</div>

```python
import numpy as np
from matplotlib import pyplot as plt
import cv2
```

<div dir="rtl" align="right">
<b>إنشاء صورة سوداء فارغة</b><br>
نستخدم <code>np.zeros()</code> لإنشاء مصفوفة مملوءة بالأصفار بأبعاد 512×512 بكسل و3 قنوات لونية. القيمة 0 تعني اللون الأسود، وهذه هي لوحة الرسم التي سنرسم عليها.
</div>

```python
img = np.zeros((512,512,3), np.uint8)
img
```

<div dir="rtl" align="right">
<b>عرض الصورة الفارغة (السوداء)</b><br>
نعرض الصورة الفارغة قبل البدء بالرسم للتأكد من إنشائها بشكل صحيح.
</div>

```python
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.show()
```

<div dir="rtl" align="right">
<b>رسم خط مستقيم</b><br>
نستخدم <code>cv2.line()</code> لرسم خط من الزاوية العلوية اليسرى (0,0) إلى الزاوية السفلية اليمنى (511,511).<br>
المعاملات: <code>(الصورة، نقطة البداية، نقطة النهاية، اللون BGR، سماكة الخط)</code>.
</div>

```python
cv2.line(img,(0,0),(511,511),(0,255,255),2)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.show()
```

```python
cv2.line(img,(0,511),(511,0),(0,255,0),5)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.show()
```

<div dir="rtl" align="right">
<b>رسم مستطيل</b><br>
نستخدم <code>cv2.rectangle()</code> لرسم مستطيل أخضر اللون.<br>
المعاملات: <code>(الصورة، الزاوية العلوية اليسرى، الزاوية السفلية اليمنى، اللون BGR، سماكة الحدود)</code>. القيمة الموجبة للسماكة ترسم الحدود فقط.
</div>

```python
cv2.rectangle(img,(100,100),(300,300),(0,255,0),10)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.show()
```

<div dir="rtl" align="right">
<b>رسم دائرة</b><br>
نستخدم <code>cv2.circle()</code> لرسم دائرة حمراء اللون.<br>
المعاملات: <code>(الصورة، إحداثيات المركز، نصف القطر، اللون BGR، سماكة الحدود)</code>. القيمة <code>-1</code> للسماكة تملأ الشكل بالكامل بالدلون.
</div>

```python
cv2.circle(img,(447,63),63,(0,0,255),3)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.show()
```

<div dir="rtl" align="right">
<b>رسم قطع ناقص (Ellipse) مملوء</b><br>
نستخدم <code>cv2.ellipse()</code> لرسم نصف قطع ناقص بنفسجي اللون مملوء بالكامل.<br>
المعاملات: <code>(الصورة، المركز، (نصف المحور الأفقي، نصف المحور الرأسي)، زاوية الدوران، زاوية البداية، زاوية النهاية، اللون، -1 للتعبئة)</code>.
</div>

```python
cv2.ellipse(img, (256,256),(95,45),0,0,180,(0,255,255),-1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.show()
```

```python
cv2.ellipse(img, (256,256),(95,45),0,0,360,(0,255,0),-1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.show()
```

<div dir="rtl" align="right">
<b>كتابة نص على الصورة</b><br>
نستخدم <code>cv2.putText()</code> لكتابة نص أبيض على الصورة.<br>
المعاملات: <code>(الصورة، النص، موضع البداية، نوع الخط، حجم الخط، اللون BGR، سماكة الخط)</code>.
</div>

```python
font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(img,'Hello_hi',(10,200),font, 2,(255,255,255),4)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.show()
```

```python
font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(img,'Hello_hi',(200,30),font, 2,(255,255,255),3)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.show()
```
