نستورد مكتبة <code>cv2</code> لمعالجة الصور، و<code>matplotlib</code> لعرضها، و<code>numpy</code> للعمليات على المصفوفات.

```python
import cv2
from  matplotlib import pyplot as plt
import numpy as np
```

<div dir="rtl" align="right">
<b>قراءة الصورة وتحويلها إلى رمادي</b><br>
نقرأ الصورة ونحولها إلى تدرج رمادي لأن عمليات العتبة (Thresholding) تعمل على صور أحادية القناة حيث تمثل كل قيمة شدة الإضاءة فقط.
</div>

```python
image = cv2.imread('flowers.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

<div dir="rtl" align="right">
<b>عرض الصورة الرمادية الأصلية</b><br>
نعرض الصورة الرمادية قبل تطبيق أي عتبة كمرجع للمقارنة مع النتائج.
</div>

```python
plt.imshow(image, cmap = 'gray', vmin=0, vmax = 255)
plt.show()
```

<div dir="rtl" align="right">
<b>رابط توثيق OpenCV للعتبة</b><br>
يحتوي هذا الرابط على الوثائق الرسمية لدوال Thresholding في OpenCV للرجوع إليها عند الحاجة.
</div>

```python
#https://docs.opencv.org/4.x/d7/d4d/tutorial_py_thresholding.html
```

<div dir="rtl" align="right">
<b>تطبيق أنواع العتبة الخمسة</b><br>
نطبق <code>cv2.threshold()</code> بخمسة أنماط مختلفة على نفس الصورة، باستخدام قيمة عتبة 127 وقيمة قصوى 255:<br><br>
— <b>THRESH_BINARY</b>: البكسلات أعلى من 127 تصبح 255 (أبيض)، وما دونها تصبح 0 (أسود).<br>
— <b>THRESH_BINARY_INV</b>: عكس السابق، البكسلات أعلى من 127 تصبح 0 والباقي 255.<br>
— <b>THRESH_TRUNC</b>: البكسلات أعلى من 127 تُقطع إلى 127، والباقي يبقى كما هو.<br>
— <b>THRESH_TOZERO</b>: البكسلات أعلى من 127 تبقى كما هي، وما دونها تصبح 0.<br>
— <b>THRESH_TOZERO_INV</b>: عكس السابق، البكسلات دون 127 تبقى والباقي يصبح 0.
</div>

```python
ret,thresh1 = cv2.threshold(image,127,255,cv2.THRESH_BINARY)
ret,thresh2 = cv2.threshold(image,127,255,cv2.THRESH_BINARY_INV)
ret,thresh3 = cv2.threshold(image,127,255,cv2.THRESH_TRUNC)
ret,thresh4 = cv2.threshold(image,127,255,cv2.THRESH_TOZERO)
ret,thresh5 = cv2.threshold(image,127,255,cv2.THRESH_TOZERO_INV)
```
