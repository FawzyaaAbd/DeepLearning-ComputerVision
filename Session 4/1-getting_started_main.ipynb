```python
from matplotlib import pyplot as plt
```

<div dir="rtl" align="right">
<b>عرض الصورة بنظام BGR</b><br>
نعرض الصورة باستخدام <code>imshow()</code>، لكن OpenCV يقرأ الصور بترتيب BGR بينما Matplotlib يتوقع RGB، لذلك ستظهر الألوان بشكل غير صحيح.
</div>

```python
plt.imshow(image)
plt.show()

# BGR  --> RGB
```

<div dir="rtl" align="right">
<b>تحويل الألوان من BGR إلى RGB وعرض الصورة</b><br>
نستخدم <code>cv2.cvtColor()</code> لتحويل الصورة من BGR إلى RGB حتى تظهر الألوان بشكل صحيح عند عرضها بـ Matplotlib.
</div>

```python
# BGR  --> RGB
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.show()
```

<div dir="rtl" align="right">
<b>تعريف دالة مخصصة لعرض الصور</b><br>
نعرّف دالة <code>imshow()</code> مخصصة تجمع عملية تحويل الألوان وعرض الصورة مع عنوان في خطوة واحدة.
</div>

```python
def imshow(title="", image=None):
    # BGR  --> RGB
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()
```

<div dir="rtl" align="right">
<b>استدعاء الدالة المخصصة لعرض الصورة</b><br>
نستدعي الدالة <code>imshow()</code> التي عرّفناها لعرض صورة الأزهار مع العنوان المحدد.
</div>

```python
imshow("display flowers", image)
```

<div dir="rtl" align="right">
<b>حفظ الصورة بصيغة JPG</b><br>
نستخدم <code>cv2.imwrite()</code> لحفظ الصورة على القرص بصيغة JPG. ترجع الدالة <code>True</code> إذا تم الحفظ بنجاح.
</div>

```python
cv2.imwrite('output.jpg',image)
```

<div dir="rtl" align="right">
<b>حفظ الصورة بصيغة PNG</b><br>
