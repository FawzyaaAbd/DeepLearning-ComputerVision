<b>استيراد مكتبة OpenCV</b><br>
نقوم باستيراد مكتبة <code>cv2</code> وهي المكتبة الرئيسية لمعالجة الصور في Python.
</div>

```python
import cv2
```

<div dir="rtl" align="right">
<b>قراءة الصورة من الملف</b><br>
نستخدم <code>cv2.imread()</code> لتحميل صورة الأزهار من القرص وتخزينها في المتغير <code>image</code> بصيغة BGR.
</div>

```python
image = cv2.imread('flowers.jpg')
```

<div dir="rtl" align="right">
<b>استيراد مكتبة Matplotlib</b><br>
نستورد مكتبة <code>matplotlib.pyplot</code> لعرض الصور داخل الـ Notebook.
</div>

```python
from matplotlib import pyplot as plt
```

<div dir="rtl" align="right">
<b>عرض الصورة الملونة بعد تحويل BGR إلى RGB</b><br>
نعرض الصورة الأصلية بعد تحويل نظام الألوان من BGR إلى RGB حتى تظهر الألوان بشكل صحيح.
</div>

```python
plt.imshow(cv2.cvtColor(image,cv2.COLOR_BGR2RGB))
plt.show
```

<div dir="rtl" align="right">
<b>تحويل الصورة إلى تدرج الرمادي (Grayscale)</b><br>
نستخدم <code>cv2.cvtColor()</code> مع الثابت <code>COLOR_BGR2GRAY</code> لتحويل الصورة الملونة إلى صورة رمادية.<br>
يتم حساب قيمة كل بكسل وفق المعادلة: <code>Gray = 0.299×R + 0.587×G + 0.114×B</code>
</div>

```python
gray = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
#Gray = 0.299 * R + 0.587 * G + 0.114 * B
```

<div dir="rtl" align="right">
<b>عرض الصورة الرمادية بالخريطة اللونية الافتراضية</b><br>
نعرض الصورة الرمادية باستخدام Matplotlib، لكن بدون تحديد الخريطة اللونية ستظهر بألوان غير صحيحة (viridis).
</div>

```python
plt.imshow(gray)
plt.show
```

<div dir="rtl" align="right">
<b>عرض الصورة الرمادية بالخريطة الصحيحة</b><br>
نعرض الصورة الرمادية مع تحديد <code>cmap="gray"</code> لضمان ظهورها بالتدرج الرمادي الصحيح، مع تثبيت نطاق القيم بين 0 و255.
</div>

```python
plt.imshow(gray, cmap ='gray', vmin=0, vmax=255)
plt.show
```

<div dir="rtl" align="right">
<b>حفظ الصورة الرمادية</b><br>
نستخدم <code>cv2.imwrite()</code> لحفظ الصورة الرمادية على القرص بصيغة JPG.
</div>

```python
cv2.imwrite('gray.jpg',gray)
```
